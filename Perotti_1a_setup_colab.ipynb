# What is Google Colab?

**Google Colaboratory (Colab)** is a free, cloud-based notebook environment provided by Google. It requires no setup, runs entirely in your browser, and provides free access to computing resources including GPUs. Notebooks you create in Colab are stored in your Google Drive and can be shared like Google Docs; a notebook opened from GitHub (like this one) is not in your Drive until you save a copy.

A notebook is code + text + output in one document. Colab runs it on Google's machines, so you need no local installation — but the machine is **temporary**: anything you do not save to Drive is gone when the session ends.

The runtime disconnects after a period of inactivity and has a maximum lifetime. When that happens, variables and files in `/content` are lost: re-run the cells (**Runtime → Run all**).

````python
# Download an input file (if not already there). This one is just a test
import os, subprocess, shlex
DEST_PATH = "/content/drive/MyDrive/dummy_text.txt"
if not os.path.exists(DEST_PATH):
    URL = "https://gist.githubusercontent.com/test1-chaliy/1023876/raw/db651f7d08f50c2ad70f24259462bb7964ba23a4/file1.txt"
    cmd = f'wget -q -O {shlex.quote(DEST_PATH)} {shlex.quote(URL)}'
    ret = subprocess.call(cmd, shell=True)
    if ret != 0:
        raise RuntimeError("Download failed. Check the URL and your connection.")
    else:
        print("Downloaded dummy_text.txt!")
else:
    print("dummy_text.txt already present:", DEST_PATH)
````

# What is a Notebook?

A **notebook** is an interactive document that combines executable code, rich text, visualizations, and media. Notebooks consist of cells, which can be either code cells (containing executable Python code) or text cells (containing formatted text using Markdown). This structure allows you to document your work alongside your code, making it ideal for data analysis, machine learning, and collaborative projects.

## Keyboard Shortcuts by Operating System

**Cmd/Ctrl + M, then X** means: press Cmd/Ctrl + M, release, then press X.

### Mac
- **Cmd + M, then A**: Insert cell above
- **Cmd + M, then B**: Insert cell below
- **Cmd + M, then M**: Convert cell to Markdown (text)
- **Cmd + M, then Y**: Convert cell to Code
- **Cmd + M, then D**: Delete current cell
- **Cmd + M, then K / J**: Move cell up / down
- **Shift + Enter**: Run cell and move to next
- **Option + Enter**: Run cell and insert new cell below
- **Ctrl + Enter**: Run cell without moving

### Windows/Linux
- **Ctrl + M, then A**: Insert cell above
- **Ctrl + M, then B**: Insert cell below
- **Ctrl + M, then M**: Convert cell to Markdown (text)
- **Ctrl + M, then Y**: Convert cell to Code
- **Ctrl + M, then D**: Delete current cell
- **Ctrl + M, then K / J**: Move cell up / down
- **Shift + Enter**: Run cell and move to next
- **Alt + Enter**: Run cell and insert new cell below
- **Ctrl + Enter**: Run cell without moving
