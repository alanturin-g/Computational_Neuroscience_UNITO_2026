## Exercise 1: List Operations

Create a list of numbers from 1 to 10. Then:
1. Use slicing to get only the even-indexed elements (0, 2, 4, ...)
2. Use a list comprehension to square each number in the original list
3. Calculate the sum of all squared numbers

**Solution**

```python
# Create list
nums = list(range(1, 11))
print("Original list:", nums)
```

```python
# 1. Even-indexed elements
even_indexed = nums[::2]
print("Even-indexed:", even_indexed)
```

```python
# 2. Square each number
squared = [x**2 for x in nums]
print("Squared:", squared)
```

```python
# 3. Sum of squared numbers
total = sum(squared)
print("Sum of squares:", total)
```

## Exercise 2: Dictionary Basics

Create a dictionary with 5 fruits as keys and their prices as values. Then:
1. Print all fruit names
2. Find the most expensive fruit using max()
3. Add 10% to all prices using a dictionary comprehension

**Solution**

```python
# Create dictionary
fruits = {'apple': 1.2, 'banana': 0.5, 'orange': 0.8, 'grape': 2.0, 'mango': 1.5}
print("Original prices:", fruits)
```

```python
# 1. Print all fruit names
print("Fruit names:", list(fruits.keys()))
```

```python
# 2. Most expensive fruit
most_expensive = max(fruits, key=fruits.get)
print(f"Most expensive: {most_expensive} at ${fruits[most_expensive]}")
```

```python
# 3. Add 10% to all prices (and enjoy python's numeric issues..)
increased_prices = {fruit: price * 1.1 for fruit, price in fruits.items()}
print("Prices after 10% increase:", increased_prices)
```

## Exercise 3: Array Creation and Basic Operations

Create a NumPy array of 10 random integers between 1 and 50. Then:
1. Find the mean, min, and max values
2. Count how many numbers are greater than 25
3. Replace all numbers less than 20 with 0

**Solution**

```python
import numpy as np

# Create random array
np.random.seed(42)
arr = np.random.randint(1, 51, 10)
print("Original array:", arr)
```

## Exercise 4: 2D Array Slicing

Create a 4x4 matrix with values from 1 to 16. Then:
1. Extract the first two rows
2. Extract the last column
3. Extract the 2x2 center submatrix (elements from rows 1-2, columns 1-2)
4. Calculate the sum of each row

**Solution**

```python
import numpy as np

# Create 4x4 matrix
matrix = np.arange(1, 17).reshape(4, 4)
print("Original matrix:")
print(matrix)
print()
```

```python
# 1. First two rows
first_two_rows = matrix[:2, :]
print("First two rows:")
print(first_two_rows)
print()
```

```python
# 2. Last column
last_column = matrix[:, -1]
print("Last column:", last_column)
print()
```

```python
# 3. Center 2x2 submatrix
center = matrix[1:3, 1:3]
print("Center 2x2 submatrix:")
print(center)
print()
```

```python
# 4. Sum of each row
row_sums = matrix.sum(axis=1)
print("Sum of each row:", row_sums)
```

## Exercise 5: Simple DataFrame Operations

Create a DataFrame with 5 students (Name, Age, Grade). Then:
1. Add a new column "Pass" that is True if Grade >= 60, False otherwise
2. Calculate the average grade
3. Filter to show only students who passed
4. Sort by Grade in descending order

**Solution**

```python
import pandas as pd

# Create DataFrame
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Eve'],
    'Age': [20, 21, 19, 22, 20],
    'Grade': [85, 55, 92, 45, 78]
}
df = pd.DataFrame(data)
print("Original DataFrame:")
print(df)
print()
```

```python
# 2. Calculate average grade
avg_grade = df['Grade'].mean()
print(f"Average grade: {avg_grade:.2f}")
print()
```

```python
# 3. Filter passing students
passing_students = df[df['Pass'] == True]
print("Students who passed:")
print(passing_students)
print()
```

```python
# 4. Sort by Grade descending
sorted_df = df.sort_values('Grade', ascending=False)
print("Sorted by Grade (descending):")
print(sorted_df)
```

## Exercise 6: Behavioural data with groupby

A participant did a task with two conditions. Each row is one trial.
Using the DataFrame below:
1. Compute the mean and standard deviation of the reaction time (`rt`) for each condition
2. Compute the accuracy (fraction of `correct` trials) for each condition
3. Remove the trials with `rt` above 1000 ms and recompute the mean RT per condition
4. Make a bar plot of the mean RT per condition (after step 3), with the standard deviation as error bars

```python
import numpy as np
import pandas as pd

np.random.seed(0)
n = 40
condition = np.tile(['congruent', 'incongruent'], n // 2)
rt = np.where(condition == 'congruent',
              np.random.normal(450, 60, n),
              np.random.normal(520, 80, n))
trials = pd.DataFrame({
    'trial': np.arange(n),
    'condition': condition,
    'rt': rt.round(),
    'correct': np.random.rand(n) > 0.15,
})
trials.loc[[5, 22], 'rt'] = [1350, 1200]  # two slow outliers
trials.head()
```

**Solution**

```python
# 1. Mean and std of RT per condition
print(trials.groupby('condition')['rt'].agg(['mean', 'std']))
```

```python
# 2. Accuracy per condition
print(trials.groupby('condition')['correct'].mean())
```

```python
# 3. Remove slow trials and recompute
clean_trials = trials[trials['rt'] <= 1000]
print(f"Removed {len(trials) - len(clean_trials)} trials")
summary = clean_trials.groupby('condition')['rt'].agg(['mean', 'std'])
print(summary)
```

```python
# 4. Bar plot with error bars
import matplotlib.pyplot as plt

plt.figure(figsize=(6, 5))
plt.bar(summary.index, summary['mean'], yerr=summary['std'], capsize=8,
        color=['skyblue', 'salmon'], edgecolor='black')
plt.ylabel('Reaction time (ms)')
plt.title('Mean RT per condition')
plt.grid(axis='y', alpha=0.3)
plt.show()
```

## Exercise 7: Multi-Plot Figure

Create a 2x2 subplot figure showing:
1. A line plot of y = x² for x from 0 to 10
2. A scatter plot of 50 random points
3. A histogram of 1000 random normal values
4. A bar plot showing values [15, 23, 18, 30, 25] for categories A-E

All plots should have appropriate titles, labels, and grids.

**Solution**

```python
import numpy as np
import matplotlib.pyplot as plt

# Solution 7: Multi-Plot Figure
np.random.seed(42)

plt.figure(figsize=(12, 10))

# 1. Line plot of y = x²
plt.subplot(2, 2, 1)
x = np.linspace(0, 10, 100)
y = x**2
plt.plot(x, y, linewidth=2, color='blue')
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Line Plot: y = x²')
plt.grid(True, alpha=0.3)

# 2. Scatter plot of 50 random points
plt.subplot(2, 2, 2)
x_scatter = np.random.randn(50)
y_scatter = np.random.randn(50)
plt.scatter(x_scatter, y_scatter, s=50, alpha=0.6, color='red')
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Scatter Plot: 50 Random Points')
plt.grid(True, alpha=0.3)

# 3. Histogram of 1000 random normal values
plt.subplot(2, 2, 3)
data_hist = np.random.randn(1000)
plt.hist(data_hist, bins=30, color='green', edgecolor='black', alpha=0.7)
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.title('Histogram: 1000 Random Normal Values')
plt.grid(axis='y', alpha=0.3)

# 4. Bar plot
plt.subplot(2, 2, 4)
categories = ['A', 'B', 'C', 'D', 'E']
values = [15, 23, 18, 30, 25]
plt.bar(categories, values, color='purple', alpha=0.7)
plt.xlabel('Category')
plt.ylabel('Value')
plt.title('Bar Plot: Categories A-E')
plt.grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.show()
```

## Exercise 8: Custom Scatter Plot

Create a scatter plot where:
1. Generate 100 random (x, y) points from a normal distribution
2. Color each point based on its distance from the origin (use sqrt(x² + y²))
3. Make point size proportional to the same distance
4. Add a colorbar, labels, title, and grid
5. Use the 'coolwarm' colormap

**Solution**

```python
import numpy as np
import matplotlib.pyplot as plt

# Solution 8: Custom Scatter Plot

np.random.seed(42)

# 1. Generate 100 random points
n = 100
x = np.random.randn(n)
y = np.random.randn(n)

# 2. Calculate distance from origin
distances = np.sqrt(x**2 + y**2)

# 3. Size proportional to distance
sizes = distances * 50  # Scale factor for visibility

# 4. Create scatter plot
plt.figure(figsize=(10, 8))
scatter = plt.scatter(x, y, c=distances, s=sizes, alpha=0.6, cmap='coolwarm')
plt.colorbar(scatter, label='Distance from Origin')
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Scatter Plot: Color and Size by Distance from Origin')
plt.grid(True, alpha=0.3)

# Add origin marker
plt.scatter(0, 0, c='black', s=200, marker='x', linewidths=3, label='Origin')
plt.legend()

plt.show()

print(f"Distance range: {distances.min():.2f} to {distances.max():.2f}")
```

## Exercise 9: Image Manipulation with NumPy Slicing

Using the image loaded in the next cell:
1. Extract only the top-left quarter of the image using slicing
2. Downsample the image by taking every other row and column (use `img[::2, ::2]`)
3. Create a vertically flipped version using slicing
4. Display all four versions (original + 3 manipulated) in a 2x2 subplot

```python
# Load the image used in this exercise
import urllib.request
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image

urllib.request.urlretrieve('https://picsum.photos/400/300', 'sample.jpg')
img = np.array(Image.open('sample.jpg'))
print(img.shape)
```

**Solution**

```python
# Solution 9: Image Manipulation with NumPy Slicing

# 1. Top-left quarter
height, width = img.shape[:2]
top_left = img[:height//2, :width//2]

# 2. Downsample (every other row and column)
downsampled = img[::2, ::2]

# 3. Vertically flipped
flipped_vertical = img[::-1, :, :]

# Display all four
plt.figure(figsize=(12, 12))

plt.subplot(2, 2, 1)
plt.imshow(img)
plt.title('Original')
plt.axis('off')

plt.subplot(2, 2, 2)
plt.imshow(top_left)
plt.title('Top-Left Quarter\n(img[:h//2, :w//2])')
plt.axis('off')

plt.subplot(2, 2, 3)
plt.imshow(downsampled)
plt.title('Downsampled\n(img[::2, ::2])')
plt.axis('off')

plt.subplot(2, 2, 4)
plt.imshow(flipped_vertical)
plt.title('Vertically Flipped\n(img[::-1, :, :])')
plt.axis('off')

plt.tight_layout()
plt.show()

print(f"Original shape: {img.shape}")
print(f"Top-left quarter: {top_left.shape}")
print(f"Downsampled: {downsampled.shape}")
```

## Exercise 10: MNIST Visualization

Using the MNIST dataset:
1. Find and display 10 examples of the digit "7" from the dataset
2. Display them in a single row (1x10 subplot)
3. Add titles showing the index of each image in the dataset
4. Notice the variation in handwriting styles!

```python
# Load MNIST (takes a minute)
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import fetch_openml

mnist = fetch_openml('mnist_784', version=1, parser='auto')
X_mnist = mnist.data.values.reshape(-1, 28, 28)
y_mnist = mnist.target.values.astype(int)
print(X_mnist.shape)
```

**Solution**

```python
# Solution 10: MNIST Visualization

# 1. Find indices where digit is 7
indices_of_7 = np.where(y_mnist == 7)[0]

# Take first 10 examples
first_10_sevens = indices_of_7[:10]

print(f"Found {len(indices_of_7)} examples of digit '7'")
print(f"Displaying first 10 at indices: {first_10_sevens}")

# 2. Display in a single row (1x10 subplot)
plt.figure(figsize=(20, 3))

for i, idx in enumerate(first_10_sevens):
    plt.subplot(1, 10, i+1)
    plt.imshow(X_mnist[idx], cmap='gray')
    plt.title(f'Index: {idx}', fontsize=10)
    plt.axis('off')

plt.suptitle('10 Examples of Handwritten Digit "7"', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()

# 4. Notice the variation!
print("\nNotice the variation in handwriting styles:")
print("- Different slants and angles")
print("- Some with horizontal bars, some without")
print("- Varying stroke thickness")
print("- Different sizes and positions within the 28x28 grid")
```
